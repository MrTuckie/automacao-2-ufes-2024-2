```python
# Exemplo 1 do Capítulo 2
# Desenhando gráficos 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Numpy - python é exclude last.
vetor = np.arange(1,101,1)
print(vetor)
```

```python

noise_0 =2*np.pi*np.random.random() 
noise_1 = 0.5*np.random.randn(100)
print(noise_0)
print(noise_1)
data = np.cos(2*np.pi*0.05*vetor + noise_0) + noise_1
fig, ax = plt.subplots()
ax.plot(vetor,data)
plt.show()
```

```python
for i in range(1,9):
    data[19+i:80+i:20] = np.nan # python slicing is start:stop:step
data[10:91:20] = [-50, 40, 30, -45, 35]
```

```python
fig,ax = plt.subplots()
ax.grid()
ax.plot(vetor,data,'.')
plt.show()
```
